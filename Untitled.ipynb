```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from pathlib import Path

RUN = "wf_2y_rth"
p = Path(f"reports/{RUN}_combined/window_pnl.csv")
wp = pd.read_csv(p)
wp["end"] = pd.to_datetime(wp["end"])
wp = wp.sort_values("end").reset_index(drop=True)

# Equity curve (test-month-only PnL aggregated across shards)
wp["equity"] = wp["pnl_sum_test_total"].cumsum()

# Drawdown stats
roll_max = wp["equity"].cummax()
dd = wp["equity"] - roll_max
wp["drawdown"] = dd
max_dd = dd.min()
max_dd_end = wp.loc[dd.idxmin(), "end"]
max_dd_start = wp.loc[: dd.idxmin(), "equity"].idxmax()
max_dd_start_date = wp.loc[max_dd_start, "end"]

# Simple outlier flag on per-window PnL
z = (wp["pnl_sum_test_total"] - wp["pnl_sum_test_total"].mean()) / (wp["pnl_sum_test_total"].std(ddof=0) or 1.0)
wp["pnl_z"] = z
flags = wp.loc[wp["pnl_z"].abs() >= 3, ["window","end","pnl_sum_test_total","trades_in_test_month_total","pnl_per_trade_test","pnl_z"]]

# Show key tables
display(wp[["window","end","pnl_sum_test_total","trades_in_test_month_total","pnl_per_trade_test","equity","drawdown"]].head())
display(wp[["window","end","pnl_sum_test_total","trades_in_test_month_total","pnl_per_trade_test"]].tail())
print(f"Max drawdown: {max_dd:.2f} from {max_dd_start_date.date()} to {max_dd_end.date()}")
display(flags)

# Plot equity + per-window bars
plt.figure(figsize=(10,4))
plt.plot(wp["end"], wp["equity"])
plt.title(f"Equity (test-month only) · {RUN}")
plt.xlabel("End month"); plt.ylabel("Equity")
plt.tight_layout(); plt.show()

plt.figure(figsize=(10,4))
plt.bar(wp["end"].astype(str), wp["pnl_sum_test_total"])
plt.title(f"Per-window PnL (test-month only) · {RUN}")
plt.xticks(rotation=75); plt.xlabel("Window end"); plt.ylabel("PnL")
plt.tight_layout(); plt.show()
```

```python
import json, glob, pandas as pd, numpy as np, os
from pathlib import Path

RUN = "wf_2y_rth"
outdir = Path(f"reports/{RUN}_combined")
outdir.mkdir(parents=True, exist_ok=True)

rows = []
for d in sorted(glob.glob(f"reports/*_{RUN}_S*of*")):
    meta_p = Path(d)/"run_meta.json"
    tr_p   = Path(d)/"trades.csv"
    if not (meta_p.exists() and tr_p.exists()):
        continue
    meta = json.loads(meta_p.read_text())
    start = meta.get("start") or meta.get("args",{}).get("start")
    end   = meta.get("end")   or meta.get("args",{}).get("end")
    test_month = (end or "")[:7]
    win = f"{(start or '')[:7]}→{test_month}"

    t = pd.read_csv(tr_p)
    # robust column picks
    sym = next((c for c in ["symbol","ticker","sym"] if c in t.columns), None)
    pnl = next((c for c in ["pnl","pnl_net","pnl_$","PNL","profit"] if c in t.columns), None)
    ent = "entry_ts" if "entry_ts" in t.columns else ("entry" if "entry" in t.columns else None)
    if not (sym and pnl and ent) or t.empty:
        continue
    t[ent] = pd.to_datetime(t[ent], errors="coerce", utc=True)
    t = t.loc[t[ent].dt.strftime("%Y-%m") == test_month].copy()
    if t.empty:
        continue
    g = t.groupby(sym, as_index=False)[pnl].sum()
    g["window"] = win
    rows.append(g)

if not rows:
    print("No per-symbol rows found."); 
else:
    ps = pd.concat(rows, ignore_index=True)
    # pivot: window x symbol PnL
    pivot = ps.pivot_table(index="window", columns=sym, values=pnl, aggfunc="sum").fillna(0).reset_index()
    pivot = pivot.sort_values("window")
    pivot.to_csv(outdir/"per_symbol_pnl_by_window.csv", index=False)
    display(pivot.head())

    # Drill into worst window by total PnL
    wp = pd.read_csv(outdir/"window_pnl.csv")
    worst_win = wp.loc[wp["pnl_sum_test_total"].idxmin(), "window"]
    print("Worst window:", worst_win)
    wrows = ps.loc[ps["window"]==worst_win]
    top_loss = wrows.sort_values(pnl).head(10)
    top_gain = wrows.sort_values(pnl, ascending=False).head(10)
    display(top_loss); display(top_gain)
```

```python
import pandas as pd, numpy as np, json, glob, os
from pathlib import Path

RUN = "wf_2y_rth"  # change if needed
base_dir = Path(f"reports/{RUN}_combined")
opt_out = base_dir / "opt_window_pnl.csv"

# 1) Load optimizer summaries across shards/windows
rows = []
for d in sorted(glob.glob(f"reports/*_{RUN}_S*of*")):
    s = Path(d) / "summary.csv"
    m = Path(d) / "run_meta.json"
    if not (s.exists() and m.exists()):
        continue
    meta = json.loads(m.read_text())
    start = meta.get("start") or meta.get("args",{}).get("start")
    end   = meta.get("end")   or meta.get("args",{}).get("end")
    win   = f"{(start or '')[:7]}→{(end or '')[:7]}"

    df = pd.read_csv(s)
    # Heuristics to find the optimizer's TEST-month aggregates:
    # try common column names; adjust 'test_pnl_cols' if needed after you print df.columns
    cols = [c.lower() for c in df.columns]
    # the presence of many rows => per-combo metrics; a single row => already "best"
    multi_rows = len(df) > 1

    # Guess the test PnL column
    test_pnl_col = None
    candidates = [
        "test_pnl", "pnl_test", "test_pnl_sum", "sum_pnl_test",
        "test_profit", "test_return", "ret_test", "pnl"  # last fallback
    ]
    for cand in candidates:
        if cand in cols:
            test_pnl_col = df.columns[cols.index(cand)]
            break
    if test_pnl_col is None:
        print(f"[WARN] Could not detect test PnL column in {s}; columns={list(df.columns)}")
        continue

    # If multiple combos present, take the row the optimizer would call "best".
    # We try common ranking fields; otherwise fall back to max(test_pnl)
    rank_fields = ["test_sharpe","sharpe_test","test_sortino","test_return","test_pnl"]
    score = None
    for rf in rank_fields:
        if rf in cols:
            score = df[df.columns[cols.index(rf)]]
            # maximize these; if you prefer Sharpe first, this order already enforces that
            best_idx = score.idxmax()
            break
    if score is None:
        # fallback: maximize test PnL
        best_idx = df[test_pnl_col].idxmax()

    best = df.loc[[best_idx]].copy()
    # Keep for window aggregation
    rows.append({
        "window": win,
        "start": start, "end": end,
        "opt_test_pnl": float(best[test_pnl_col].values[0]),
        "_multi_rows": multi_rows,
        # keep param columns for possible global-param step
        "params": best.to_dict(orient="records")[0]
    })

opt = pd.DataFrame(rows)
if opt.empty:
    raise RuntimeError("No optimizer summaries found. Check RUN and that summary.csv exists per shard.")

# 2) Aggregate across shards per window
opt_win = opt.groupby(["window","start","end"], dropna=False)["opt_test_pnl"].sum().reset_index()
opt_win = opt_win.sort_values("end")
opt_win.to_csv(opt_out, index=False)
display(opt_win.head())
print(f"Wrote: {opt_out}")

# 3) Compare baseline vs optimizer per-window PnL
base_win = pd.read_csv(base_dir / "window_pnl.csv")
base_win = base_win.sort_values("end")
cmp = base_win.merge(opt_win, on=["window","start","end"], how="left")
cmp["lift_vs_base"] = cmp["opt_test_pnl"] - cmp["pnl_sum_test_total"]
display(cmp[["window","end","pnl_sum_test_total","opt_test_pnl","lift_vs_base"]].head(10))
display(cmp.tail(5))

# Summaries
print("Baseline total PnL:", round(base_win["pnl_sum_test_total"].sum(), 2))
print("Optimized total PnL:", round(opt_win["opt_test_pnl"].sum(), 2))
print("Lift:", round(opt_win["opt_test_pnl"].sum() - base_win["pnl_sum_test_total"].sum(), 2))

# Did we have per-combo summaries?
any_multi = bool(opt["_multi_rows"].any())
print("Optimizer summaries contain per-combo rows?", any_multi)
```

```python
import pandas as pd, numpy as np, json, glob
from pathlib import Path

RUN = "wf_2y_rth"  # change if your run-tag differs
base_dir = Path(f"reports/{RUN}_combined")
opt_out = base_dir / "opt_window_pnl.csv"

rows = []
for d in sorted(glob.glob(f"reports/*_{RUN}_S*of*")):
    s = Path(d) / "summary.csv"
    m = Path(d) / "run_meta.json"
    if not (s.exists() and m.exists()):
        continue

    meta = json.loads(m.read_text())
    start = meta.get("start") or meta.get("args",{}).get("start")
    end   = meta.get("end")   or meta.get("args",{}).get("end")
    win   = f"{(start or '')[:7]}→{(end or '')[:7]}"

    df = pd.read_csv(s)
    if df.empty:
        continue

    cols_l = {c.lower(): c for c in df.columns}
    # your files have test_total_pnl / test_trades
    test_pnl_col   = cols_l.get("test_total_pnl") or cols_l.get("test_pnl") or cols_l.get("pnl") or None
    test_trd_col   = cols_l.get("test_trades")    or None
    best_params_col= cols_l.get("best_params")    or None

    if test_pnl_col is None:
        print(f"[WARN] No recognizable test PnL in {s}; cols={list(df.columns)}")
        continue

    # summaries are already 1 row = best; if many, take max by test pnl
    row = df.loc[[df[test_pnl_col].idxmax()]] if len(df) > 1 else df.iloc[[0]]

    rows.append({
        "window": win, "start": start, "end": end,
        "opt_test_pnl": float(row[test_pnl_col].values[0]),
        "opt_test_trades": int(row[test_trd_col].values[0]) if test_trd_col else np.nan,
        "best_params": row[best_params_col].values[0] if best_params_col else None,
        "run_dir": str(d),
    })

opt = pd.DataFrame(rows)
if opt.empty:
    raise RuntimeError("No optimizer summaries parsed. Check RUN or a sample summary.csv to confirm column names.")

# per-window sum across shards
opt_win = opt.groupby(["window","start","end"], dropna=False)[["opt_test_pnl","opt_test_trades"]].sum(min_count=1).reset_index()
opt_win = opt_win.sort_values("end")
opt_win.to_csv(opt_out, index=False)

# compare to baseline (test-month trades)
base_win = pd.read_csv(base_dir / "window_pnl.csv").sort_values("end")
cmp = base_win.merge(opt_win, on=["window","start","end"], how="left")
cmp["lift_vs_base"] = cmp["opt_test_pnl"] - cmp["pnl_sum_test_total"]

display(cmp[["window","end","pnl_sum_test_total","opt_test_pnl","lift_vs_base"]].head(12))
print("Baseline total:", round(base_win["pnl_sum_test_total"].sum(), 2))
print("Optimized total:", round(opt_win["opt_test_pnl"].sum(), 2))
print("Lift:", round(opt_win["opt_test_pnl"].sum() - base_win["pnl_sum_test_total"].sum(), 2))
print(f"Wrote: {opt_out}")
```

```python
import ast, json
from collections import Counter

# Parse 'best_params' per shard/window and count frequency of param combos
def parse_params(v):
    if v is None or (isinstance(v, float) and np.isnan(v)): return None
    if isinstance(v, dict): return v
    if isinstance(v, str):
        v = v.strip()
        # try JSON then Python literal
        try: return json.loads(v)
        except Exception:
            try: return ast.literal_eval(v)
            except Exception: return None
    return None

opt["params_obj"] = opt["best_params"].apply(parse_params)
# Create a canonical key for counting (sorted items)
def keyify(d):
    if not isinstance(d, dict): return None
    return "|".join(f"{k}={d[k]}" for k in sorted(d.keys()))
opt["param_key"] = opt["params_obj"].apply(keyify)

freq = Counter([k for k in opt["param_key"] if k])
top = pd.DataFrame(freq.most_common(10), columns=["param_key","wins"])
display(top)

# (Optional) show a sample of windows where the top param was used
if not top.empty:
    k0 = top.iloc[0]["param_key"]
    sample = opt.loc[opt["param_key"]==k0, ["window","opt_test_pnl","opt_test_trades","run_dir"]].head(10)
    print("Top param used in windows:")
    display(sample)
```
